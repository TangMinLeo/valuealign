```python
clean_file = './clean_data_with_category_delete_none.csv'

value_file = './person_value_vector_988.jsonl'

import pandas as pd
import json

df = pd.read_csv(clean_file)

uuid_list = []
with open(value_file, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():  # 确保行不为空
            data = json.loads(line.strip())
            uuid_list.append(list(data.keys())[0])  # 获取每行的第一个键，即uuid

print(uuid_list)
print(len(uuid_list))
uuid_set = set(uuid_list)

filtered = df[df["uuid"].isin(uuid_set)]
# print(filtered)
print(len(filtered))

filtered.to_csv('clean_data_with_category_delete_none_988.csv', index=False)

```
